## Bonus

You work at the customer service department and you want to know which months had the highest number of complaints by policy type category. Create a summary table showing the number of complaints by policy type and month.
Show it in a long format table.

*In data analysis, a long format table is a way of structuring data in which each observation or measurement is stored in a separate row of the table. The key characteristic of a long format table is that each column represents a single variable, and each row represents a single observation of that variable.*

*More information about long and wide format tables here: https://www.statology.org/long-vs-wide-data/*

```python
# Your code goes here
marketing_customer_analysis[
    ["policy_type", "month", "number_of_open_complaints"]
].head()
```

```python
marketing_customer_analysis["has_complaint"] = (
    marketing_customer_analysis["number_of_open_complaints"] > 0
)

complaints_by_policy_month = marketing_customer_analysis.pivot_table(
    index=["policy_type", "month"],
    values="has_complaint",
    aggfunc="sum"
).reset_index()

complaints_by_policy_month = complaints_by_policy_month.rename(
    columns={"has_complaint": "number_of_complaints"}
)

complaints_by_policy_month
```

```python
complaints_by_policy_month.sort_values(
    by="number_of_complaints",
    ascending=False
)
```

The results show that Personal Auto had the highest number of customers with open complaints in both months. Month 1 had more customers with complaints than month 2 across all three policy types. Special Auto had the lowest numbers, with 51 customers in month 1 and 50 in month 2. This suggests that customer complaints were most common among Personal Auto customers during month 1.
